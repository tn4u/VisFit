```python
# Stdlib only: dependency installation must precede all benchmark imports.
import os, sys, json, re, subprocess, hashlib, importlib, importlib.metadata, importlib.util, importlib.machinery
import traceback, time, uuid, shutil
from pathlib import Path

VERIFIED_RUN_ID = CFG['run_id']
ENCODER_ID_LOCK = CFG['encoder_id']
ENCODER_REVISION_LOCK = CFG['encoder_revision']
BOOTSTRAP_PLAN = {'transformers':'4.57.1','tokenizers':'0.22.1','huggingface-hub':'0.36.0',
                  'faiss-cpu':'1.11.0','ftfy':'6.3.1'}
BOOTSTRAP_PRESERVE = ('torch','torchvision','numpy','pandas','Pillow')
PACKAGE_MODULES = {'torch':'torch','torchvision':'torchvision','transformers':'transformers',
    'tokenizers':'tokenizers','huggingface-hub':'huggingface_hub','numpy':'numpy','pandas':'pandas',
    'Pillow':'PIL','kaggle':'kaggle','faiss-cpu':'faiss','safetensors':'safetensors','scipy':'scipy',
    'scikit-learn':'sklearn','librosa':'librosa','numba':'numba','soxr':'soxr','ftfy':'ftfy'}
ENV_ROOT = Path(CFG['environment_root'])
BOOTSTRAP_POLICY_ID = hashlib.sha256(json.dumps({'version':'preparation-1.0.0','plan':BOOTSTRAP_PLAN,'install_kaggle_only_if_missing':True},sort_keys=True).encode()).hexdigest()
KERNEL_ID = str(os.getpid())
BOOTSTRAP_READY = False
ENVIRONMENT_PREFLIGHT = None

class BootstrapRestartRequired(RuntimeError):
    pass

class EnvironmentPreflightBlocked(RuntimeError):
    pass

def redact_environment_text(value):
    text=str(value)
    for key,secret in os.environ.items():
        if len(secret)>=6 and re.search(r'(TOKEN|PASSWORD|SECRET|API_KEY|ACCESS_KEY)',key,re.I):
            text=text.replace(secret,'<redacted>')
    text=re.sub(r'(?i)(Bearer\s+)[A-Za-z0-9_.-]+',r'\1<redacted>',text)
    text=re.sub(r'(https?://)[^\s/@:]+:[^\s/@]+@',r'\1<redacted>@',text)
    text=re.sub(r'(?i)([?&](?:token|key|sig|signature|x-amz-signature|access_token|auth)=)[^\s&\"\']+',r'\1<redacted>',text)
    return text

def environment_write(path, value):
    path=Path(path);path.parent.mkdir(parents=True,exist_ok=True)
    raw=json.dumps(value,ensure_ascii=False,indent=2,allow_nan=False)
    raw=redact_environment_text(raw)
    temporary=path.with_name(path.name+'.'+uuid.uuid4().hex+'.tmp');temporary.write_text(raw,encoding='utf-8')
    os.replace(temporary,path)

def package_snapshot():
    result={}
    for distribution,module_name in PACKAGE_MODULES.items():
        row={}
        try:
            dist=importlib.metadata.distribution(distribution)
            row.update(disk_version=dist.version,distribution_root=str(dist.locate_file('')),
                       requirements=dist.requires or [])
        except importlib.metadata.PackageNotFoundError:row['disk_version']=None
        try:
            spec=importlib.machinery.PathFinder.find_spec(module_name,sys.path)
            row['disk_spec_origin']=str(spec.origin) if spec else None
        except Exception:row['spec_traceback']=traceback.format_exc()
        cached=sys.modules.get(module_name)
        if cached is not None:
            row['loaded_version']=cached.__dict__.get('__version__') or cached.__dict__.get('_objects',{}).get('__version__')
            row['loaded_file']=cached.__dict__.get('__file__')
        row['already_loaded']=cached is not None
        result[distribution]=row
    return result

def loaded_distribution_snapshot():
    # Include transitive dependencies already imported by Colab/IPython, without importing them here.
    module_roots={name.split('.')[0] for name in sys.modules}
    mapping=importlib.metadata.packages_distributions()
    result={}
    for module in module_roots:
        for distribution in mapping.get(module,[]):
            try:result[distribution]=importlib.metadata.version(distribution)
            except importlib.metadata.PackageNotFoundError:pass
    return result

def shadowing_findings(snapshot):
    found=[]
    for name,row in snapshot.items():
        origin=row.get('disk_spec_origin')
        if origin and row.get('distribution_root'):
            path=Path(origin).resolve();installed=Path(row['distribution_root']).resolve()
            if installed not in path.parents:
                found.append({'package':name,'origin':str(path),'distribution_root':str(installed),
                              'reason':'import origin nằm ngoài distribution; kiểm tra file/thư mục local shadow package'})
    return found

def classify_pip_check(text):
    canonical=lambda x:re.sub(r'[-_.]+','-',x).lower()
    closure=set();pending=list(BOOTSTRAP_PLAN)+list(BOOTSTRAP_PRESERVE)+['safetensors']
    while pending:
        name=canonical(pending.pop())
        if name in closure:continue
        closure.add(name)
        try:requirements=importlib.metadata.requires(name) or []
        except importlib.metadata.PackageNotFoundError:continue
        for req in requirements:
            if 'extra ==' in req or 'extra==' in req:continue
            match=re.match(r'([A-Za-z0-9_.-]+)',req)
            if match:pending.append(match.group(1))
    optional={'scikit-learn','scipy','librosa','numba','soxr','torchvision'}
    rows=[]
    for line in text.splitlines():
        if not line.strip() or line.startswith('WARNING:') or line.strip()=='No broken requirements found.':continue
        owner=canonical(line.split()[0])
        group='benchmark_dependency' if owner in closure else 'optional_import_chain' if owner in optional else 'unrelated_installed_package'
        rows.append({'message':line,'owner':owner,'classification':group,
                     'blocking_declared_conflict':group=='benchmark_dependency'})
    return rows

def request_single_restart(reason):
    marker=ENV_ROOT/'bootstrap_state.json';state=json.loads(marker.read_text()) if marker.exists() else {}
    if state.get('restart_kernel')==KERNEL_ID:
        raise BootstrapRestartRequired('Cần restart runtime MỘT LẦN: Runtime → Restart session, rồi chạy lại từ Bootstrap. Không chạy tiếp các cell sau. Báo cáo: '+str(ENV_ROOT))
    if state.get('restart_count',0)>=1:
        raise EnvironmentPreflightBlocked('Đã yêu cầu restart một lần nhưng môi trường vẫn chưa đạt. Dừng để đọc environment_preflight.json; không tự cài/restart lặp.')
    state.update(policy=BOOTSTRAP_POLICY_ID,restart_count=1,restart_kernel=KERNEL_ID,restart_reason=reason)
    environment_write(marker,state)
    raise BootstrapRestartRequired('Cần restart runtime MỘT LẦN vì '+reason+'. Runtime → Restart session, rồi chạy lại từ Bootstrap; không chạy tiếp. Báo cáo: '+str(ENV_ROOT))

def bootstrap_environment():
    global BOOTSTRAP_READY
    BOOTSTRAP_READY=False
    ENV_ROOT.mkdir(parents=True,exist_ok=True)
    report={'stage':'bootstrap','kernel_id':KERNEL_ID,'python':sys.version,'executable':sys.executable,
            'plan':BOOTSTRAP_PLAN,'preserve':list(BOOTSTRAP_PRESERVE),'status':'blocked_environment'}
    try:
        marker=ENV_ROOT/'bootstrap_state.json';state=json.loads(marker.read_text()) if marker.exists() else {}
        if state and state.get('policy')!=BOOTSTRAP_POLICY_ID:
            raise EnvironmentPreflightBlocked('Bootstrap policy đã đổi; dùng runtime sạch hoặc thư mục/run mới, không ghi đè trạng thái cũ.')
        before=package_snapshot();loaded_before=loaded_distribution_snapshot()
        report.update(before=before,loaded_distributions_before=loaded_before,shadowing=shadowing_findings(before))
        if report['shadowing']:
            raise EnvironmentPreflightBlocked('Có module local shadow package: '+', '.join(x['package'] for x in report['shadowing'])+'. Đổi tên/chuyển file local sau khi xem báo cáo, không tự xóa.')
        if state.get('restart_kernel')==KERNEL_ID:
            request_single_restart(state.get('restart_reason','package đã thay đổi nhưng kernel cũ còn module'))
        targets=[name+'=='+version for name,version in BOOTSTRAP_PLAN.items() if before.get(name,{}).get('disk_version')!=version]
        if before['kaggle']['disk_version'] is None:targets.append('kaggle')
        # Chỉ cài scientific packages nếu thực sự thiếu; không ép phiên bản cũ lên Colab mới.
        for package,spec in [('numpy','numpy>=1.26,<3'),('pandas','pandas>=2.1,<4'),('Pillow','Pillow>=10,<13')]:
            if before[package]['disk_version'] is None:targets.append(spec)
        if before['torch']['disk_version'] is None:
            raise EnvironmentPreflightBlocked('Torch không tồn tại trong runtime; không tự thay torch/CUDA. Chọn runtime Colab chuẩn rồi kiểm tra lại.')
        if targets:
            if state.get('install_count',0)>=1:
                raise EnvironmentPreflightBlocked('Đã thử cài một lần nhưng phiên bản vẫn khác plan; xem báo cáo. Không tự reinstall hàng loạt.')
            constraints=ENV_ROOT/'preserve-colab-constraints.txt'
            constraints.write_text('\n'.join(name+'=='+before[name]['disk_version'] for name in BOOTSTRAP_PRESERVE if before[name]['disk_version'])+'\n')
            state.update(policy=BOOTSTRAP_POLICY_ID,install_count=1)
            environment_write(marker,state)
            command=[sys.executable,'-m','pip','install','--disable-pip-version-check','--quiet','--index-url','https://pypi.org/simple','-c',str(constraints),*targets]
            process=subprocess.run(command,capture_output=True,text=True,timeout=600)
            report['pip_install']={'requested':targets,'returncode':process.returncode,
                                   'stdout':process.stdout,'stderr':process.stderr}
            if process.returncode:raise EnvironmentPreflightBlocked('Cài dependency có mục tiêu thất bại; không tiếp tục. Đọc bootstrap_report.json.')
            importlib.invalidate_caches()
        after=package_snapshot();report['after']=after
        for name in BOOTSTRAP_PRESERVE:
            if before[name]['disk_version'] is not None and before[name]['disk_version']!=after[name]['disk_version']:
                raise EnvironmentPreflightBlocked('Package cần giữ đã thay đổi ngoài dự kiến: '+name)
        changed_loaded=[name for name in after if before[name].get('already_loaded') and before[name]['disk_version']!=after[name]['disk_version']]
        stale=[name for name,row in after.items() if (row.get('loaded_version') and row['loaded_version']!=row['disk_version']) or (row.get('loaded_file') and row.get('disk_spec_origin') and Path(row['loaded_file']).resolve()!=Path(row['disk_spec_origin']).resolve())]
        loaded_after=loaded_distribution_snapshot();report['loaded_distributions_after']=loaded_after
        transitive_changed=[name for name,version in loaded_before.items() if loaded_after.get(name)!=version]
        changed_loaded+=transitive_changed
        if changed_loaded or stale:request_single_restart('disk và module trong kernel khác nhau: '+', '.join(sorted(set(changed_loaded+stale))))
        state.update(policy=BOOTSTRAP_POLICY_ID,ready_kernel=KERNEL_ID,restart_observed=bool(state.get('restart_kernel') and state['restart_kernel']!=KERNEL_ID))
        environment_write(marker,state);report['status']='ready';BOOTSTRAP_READY=True
        print('Bootstrap đạt; giữ nguyên torch/CUDA và scientific packages đang có. Chạy environment preflight tiếp theo.')
    except Exception:
        report['traceback']=traceback.format_exc();raise
    finally:
        environment_write(ENV_ROOT/'bootstrap_report.json',report)
        if report['status']!='ready':
            environment_write(ENV_ROOT/'environment_preflight.json',dict(report,stage='bootstrap',benchmark_executed=False))
            environment_write(ENV_ROOT/'execution_status.json',{'run_id':VERIFIED_RUN_ID,'stage':'bootstrap','status':'blocked_environment','preparation_executed':False})

bootstrap_environment()
```

```python
# Run immediately after bootstrap, before Drive mount, dataset download or image audit.
ENV_IMPORT_PROBE = r'''
import sys,json,traceback,importlib,importlib.util,importlib.metadata
report={'tests':{},'optional_imports':{},'modules':{}}
def attempt(name,code,target):
 try:
  exec(code,globals());target[name]={'status':'passed'}
 except Exception as exc:
  target[name]={'status':'failed','exception_type':type(exc).__name__,'traceback':traceback.format_exc()}
for name,code in [
 ('GenerationMixin','from transformers.generation import GenerationMixin'),
 ('generation_utils_direct','from transformers.generation.utils import GenerationMixin'),
 ('CLIP_imports','from transformers import CLIPModel, CLIPVisionModel, CLIPTextModel, CLIPTokenizer, CLIPImageProcessor'),
 ('numpy_pandas_PIL','import numpy; import pandas; from PIL import Image'),
 ('torch_cpu','import torch; a=torch.arange(9,dtype=torch.float32).reshape(3,3); assert torch.isfinite(a@a.T).all()')]:
 attempt(name,code,report['tests'])
for module in ('torchvision','scipy','sklearn','librosa','numba','soxr'):
 if importlib.util.find_spec(module) is not None:attempt(module,'import '+module,report['optional_imports'])
try:
 from transformers.utils import import_utils
 report['torch_available_in_transformers']=bool(import_utils.is_torch_available())
 report['generation_export_registered']='GenerationMixin' in getattr(sys.modules.get('transformers.generation'),'_class_to_module',{})
except Exception:report['availability_traceback']=traceback.format_exc()
for module in ('torch','torchvision','transformers','transformers.generation','tokenizers','huggingface_hub','numpy','pandas','PIL'):
 obj=sys.modules.get(module)
 if obj is not None:report['modules'][module]={'file':getattr(obj,'__file__',None),'version':getattr(obj,'__version__',None),'class':type(obj).__name__}
report['passed']=all(row['status']=='passed' for row in report['tests'].values())
print('FASHIONIQ_ENV_JSON:'+json.dumps(report))
'''
FAISS_ENV_PROBE = r'''
import json,traceback
report={}
try:
 import numpy as np,faiss
 g=np.array([[1,0],[0,1],[-1,0]],dtype=np.float32);q=np.array([[1,0]],dtype=np.float32)
 index=faiss.IndexFlatIP(2);index.add(g);scores,ids=index.search(q,3)
 assert ids.tolist()==[[0,1,2]] and np.allclose(scores,[[1,0,-1]])
 report={'status':'passed','module_file':faiss.__file__,'module_version':getattr(faiss,'__version__',None),'exact_ids':ids.tolist()}
except Exception:report={'status':'failed','traceback':traceback.format_exc()}
print('FASHIONIQ_ENV_JSON:'+json.dumps(report))
'''

def run_environment_child(code, timeout=180):
    result=subprocess.run([sys.executable,'-c',code],text=True,capture_output=True,timeout=timeout)
    lines=[line[len('FASHIONIQ_ENV_JSON:'):] for line in result.stdout.splitlines() if line.startswith('FASHIONIQ_ENV_JSON:')]
    if lines:
        value=json.loads(lines[-1]);value['process_returncode']=result.returncode
    else:value={'status':'failed','passed':False,'process_returncode':result.returncode}
    value['stderr']=result.stderr
    if not lines:value['stdout']=result.stdout
    return value

def classify_environment_failure(exc):
    message=str(exc).lower()
    if isinstance(exc,(ImportError,ModuleNotFoundError)) or any(s in message for s in ('numpy.dtype size','operator torchvision::','undefined symbol','generationmixin')):
        return 'dependency_import'
    if any(s in message for s in ('401','403','unauthorized','gated repo','not a valid model identifier')):return 'model_access'
    if any(s in message for s in ('connection','timed out','offline','network','couldn\'t connect','name resolution','localentrynotfound','ssl','certificate')):return 'network_or_weight_download'
    return 'environment_or_forward'

def run_environment_preflight():
    global ENVIRONMENT_PREFLIGHT, _PREFLIGHT_ENCODER_BUNDLE
    report={'schema_version':'environment-preflight-v1','kernel_id':KERNEL_ID,'run_id':VERIFIED_RUN_ID,
            'status':'blocked_environment','python':sys.version,'executable':sys.executable,
            'package_snapshot':package_snapshot(),'shadowing':shadowing_findings(package_snapshot()),
            'tests':{},'cuda_test':{'status':'skipped','reason':'CUDA not available until verified'},
            'encoder_lock':{'id':ENCODER_ID_LOCK,'revision':ENCODER_REVISION_LOCK},'benchmark_executed':False}
    _PREFLIGHT_ENCODER_BUNDLE=None
    try:
        if not BOOTSTRAP_READY:raise EnvironmentPreflightBlocked('Bootstrap chưa đạt; không chạy preflight/dataset.')
        if report['shadowing']:
            report['failure_kind']='dependency_shadowing'
            raise EnvironmentPreflightBlocked('Module shadowing xuất hiện sau Bootstrap; xem package_snapshot trước khi tiếp tục.')
        pip=subprocess.run([sys.executable,'-m','pip','check'],text=True,capture_output=True,timeout=120)
        report['pip_check']={'returncode':pip.returncode,'stdout':pip.stdout,'stderr':pip.stderr,
                             'conflicts':classify_pip_check(pip.stdout)}
        fresh=run_environment_child(ENV_IMPORT_PROBE);report['fresh_process']=fresh
        faiss=run_environment_child(FAISS_ENV_PROBE);report['faiss_process']=faiss
        # Run the same import probes in the kernel to catch lazy-module/disk mismatches.
        captured=[]
        def capture_print(*args,**kwargs):captured.append(' '.join(map(str,args)))
        kernel_namespace={'print':capture_print}
        exec(ENV_IMPORT_PROBE,kernel_namespace)
        kernel=json.loads(next(line[len('FASHIONIQ_ENV_JSON:'):] for line in reversed(captured) if line.startswith('FASHIONIQ_ENV_JSON:')))
        report['kernel_process']=kernel
        if fresh.get('passed') and not kernel.get('passed'):
            report['failure_kind']='disk_kernel_mismatch'
            request_single_restart('import ở subprocess mới đạt nhưng kernel hiện tại lỗi lazy module')
        if not fresh.get('passed') or not kernel.get('passed') or faiss.get('status')!='passed':
            report['failure_kind']='dependency_import'
            raise EnvironmentPreflightBlocked('Import/FAISS preflight thất bại; đọc đầy đủ fresh_process/kernel_process và traceback trực tiếp trong environment_preflight.json.')
        conflicts=report['pip_check']['conflicts']
        if any(row['blocking_declared_conflict'] for row in conflicts):
            report['failure_kind']='dependency_conflict'
            raise EnvironmentPreflightBlocked('pip check có xung đột thuộc dependency benchmark; không tự đổi hàng loạt. Xem phân loại trong JSON.')
        import torch, numpy as np, pandas as pd
        from PIL import Image
        from transformers.generation import GenerationMixin
        from transformers import CLIPModel,CLIPVisionModel,CLIPTextModel,CLIPTokenizer,CLIPImageProcessor
        cpu=torch.arange(16,dtype=torch.float32).reshape(4,4)
        if not torch.isfinite(cpu@cpu.T).all():raise AssertionError('CPU matmul NaN/Inf')
        report['tests']['torch_cpu']={'status':'passed','dtype':str(cpu.dtype)}
        device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        if device.type=='cuda':
            tensor=cpu.to(device);value=tensor@tensor.T;torch.cuda.synchronize()
            if not torch.allclose(value.cpu(),cpu@cpu.T):raise AssertionError('CUDA matmul mismatch')
            report['cuda_test']={'status':'passed','device':torch.cuda.get_device_name(device),'cuda':torch.version.cuda}
        else:report['cuda_test']={'status':'skipped','reason':'No CUDA in this actual environment; CPU tensor test passed.'}
        torch.backends.cuda.matmul.allow_tf32=False;torch.backends.cudnn.allow_tf32=False;torch.set_float32_matmul_precision('highest')
        report['stage']='pretrained_encoder_load'
        model=CLIPModel.from_pretrained(ENCODER_ID_LOCK,revision=ENCODER_REVISION_LOCK,attn_implementation='eager').float().to(device).eval().requires_grad_(False)
        tokenizer=CLIPTokenizer.from_pretrained(ENCODER_ID_LOCK,revision=ENCODER_REVISION_LOCK)
        processor=CLIPImageProcessor.from_pretrained(ENCODER_ID_LOCK,revision=ENCODER_REVISION_LOCK)
        inputs=tokenizer(['a blue shirt',' and longer sleeves'],padding='max_length',max_length=77,truncation=True,return_tensors='pt')
        pixels=processor(images=[Image.new('RGB',(224,224),(30,80,150)),Image.new('RGB',(256,192),(150,90,40))],return_tensors='pt')['pixel_values'].float().to(device)
        report['stage']='pretrained_encoder_forward'
        with torch.inference_mode():
            vision=model.vision_model(pixel_values=pixels);text=model.text_model(**{k:v.to(device) for k,v in inputs.items()})
            projected=model.visual_projection(vision.pooler_output);gallery=torch.nn.functional.normalize(projected,dim=-1)
            api=model.get_image_features(pixel_values=pixels);api=api.pooler_output if not isinstance(api,torch.Tensor) else api
        expected={'reference':(2,50,768),'text':(2,77,512),'pool':(2,768),'projection':(2,512),'gallery':(2,512)}
        arrays={'reference':vision.last_hidden_state,'text':text.last_hidden_state,'pool':vision.pooler_output,'projection':projected,'gallery':gallery}
        for name,value in arrays.items():
            if tuple(value.shape)!=expected[name] or value.dtype!=torch.float32 or not torch.isfinite(value).all():
                raise AssertionError('Fashion-CLIP shape/dtype/finite failed: '+name)
            report['tests'][name]={'status':'passed','shape':list(value.shape),'dtype':str(value.dtype)}
        if not torch.allclose(api,projected,atol=1e-5,rtol=1e-5):raise AssertionError('get_image_features vs explicit visual projection mismatch')
        if not torch.allclose(gallery.norm(dim=-1),torch.ones(2,device=device),atol=1e-5):raise AssertionError('Gallery L2 norm mismatch')
        report['tests']['projection_L2']={'status':'passed','max_api_error':float((api-projected).abs().max().cpu())}
        report['tests']['GenerationMixin_and_CLIP_imports']={'status':'passed','note':'real imports; no monkey patch'}
        report['status']='passed';report['stage']='complete';report['package_snapshot_after']=package_snapshot()
        _PREFLIGHT_ENCODER_BUNDLE=(model,tokenizer,processor,device,ENCODER_ID_LOCK,ENCODER_REVISION_LOCK)
        print('Environment preflight đạt: imports, CPU/FAISS và Fashion-CLIP thực tại revision khóa đã forward trên input tổng hợp.')
        print('CUDA:',report['cuda_test']['status'],'| pip conflicts không liên quan:',sum(row['classification']=='unrelated_installed_package' for row in conflicts))
    except Exception as exc:
        report['failure_kind']=report.get('failure_kind') or classify_environment_failure(exc)
        report['traceback']=traceback.format_exc()
        report['exception_chain']=[];current=exc;visited=set()
        while current is not None and id(current) not in visited:
            visited.add(id(current));report['exception_chain'].append({'type':type(current).__name__,'message':str(current),
                'traceback':''.join(traceback.format_exception(type(current),current,current.__traceback__))})
            current=current.__cause__ or current.__context__
        if isinstance(exc,BootstrapRestartRequired):raise
        raise EnvironmentPreflightBlocked('DỪNG tại environment preflight ('+report['failure_kind']+'). Chưa tải dataset/audit/encode. Đọc '+str(ENV_ROOT/'environment_preflight.json')) from exc
    finally:
        ENVIRONMENT_PREFLIGHT=report
        environment_write(ENV_ROOT/'environment_preflight.json',report)
        environment_write(ENV_ROOT/'execution_status.json',{'run_id':VERIFIED_RUN_ID,'stage':'environment_preflight','status':'ready' if report['status']=='passed' else 'blocked_environment','preparation_executed':False})
        print('Environment report:',ENV_ROOT/'environment_preflight.json')
    return report

ENVIRONMENT_PREFLIGHT = run_environment_preflight()
```

## 2. Định nghĩa protocol, kiểm tra và feature store

```python
DVR_HIDDEN=256
DVR_ALPHA=1.0

class XAAFusionModule(nn.Module):
    def __init__(self, text_dim=512, img_dim=768, embed_dim=512, num_heads=8, dropout=0.1):
        super().__init__()
        self.ref_proj = nn.Linear(img_dim, embed_dim)
        self.ref_ln   = nn.LayerNorm(embed_dim)
        self.attn = nn.MultiheadAttention(embed_dim=embed_dim, num_heads=num_heads,
                                          batch_first=True, dropout=dropout)
        self.out_ln = nn.LayerNorm(embed_dim)
        self.logit_scale = nn.Parameter(torch.tensor(math.log(1 / 0.07)))

    def forward(self, text_hidden, text_mask, reference_hidden):
        ref = self.ref_ln(self.ref_proj(reference_hidden))              # [B,50,512]
        att, _ = self.attn(query=text_hidden, key=ref, value=ref)       # [B,77,512]
        x = self.out_ln(text_hidden + att)                              # residual + LN
        m = text_mask.unsqueeze(-1)                                     # [B,77,1]
        pooled = (x * m).sum(1) / m.sum(1).clamp(min=1.0)               # masked mean
        return F.normalize(pooled, dim=-1)

class ModifierGuidedRefinement(nn.Module):
    def __init__(self, dim=512, hidden=DVR_HIDDEN, alpha=DVR_ALPHA):
        super().__init__()
        self.alpha = float(alpha)
        self.v_proj = nn.Linear(dim, hidden)       # W_v
        self.t_proj = nn.Linear(dim, hidden)       # W_t
        self.act = nn.GELU()
        self.out = nn.Linear(hidden, 1)            # gate head
        nn.init.zeros_(self.out.weight); nn.init.zeros_(self.out.bias)   # identity tại khởi tạo

    def forward(self, v, text_hidden, text_mask):
        # v: [B,N,D] | text_hidden: [B,L,D] | text_mask: [B,L] (1 = token thật)
        m = text_mask.unsqueeze(-1).to(text_hidden.dtype)
        t = (text_hidden * m).sum(1) / m.sum(1).clamp(min=1.0)                  # masked mean, bỏ padding   [B,D]
        h = self.act(self.v_proj(v) + self.t_proj(t).unsqueeze(1))              # [B,N,hidden]
        gate = 1.0 + self.alpha * torch.tanh(self.out(h).squeeze(-1))          # [B,N]  ∈ (1-α, 1+α)
        return v * gate.unsqueeze(-1), gate

class SelfGuidedRefinement(nn.Module):
    def __init__(self, dim=512, hidden=DVR_HIDDEN, alpha=DVR_ALPHA):
        super().__init__()
        self.alpha = float(alpha)
        self.v_proj = nn.Linear(dim, hidden)       # W_v
        self.c_proj = nn.Linear(dim, hidden)       # W_c (ngữ cảnh thị giác toàn ảnh)
        self.act = nn.GELU()
        self.out = nn.Linear(hidden, 1)
        nn.init.zeros_(self.out.weight); nn.init.zeros_(self.out.bias)

    def forward(self, v):                           # CHỈ nhận image tokens — không có đường vào cho text
        ctx = v.mean(dim=1)                                                     # [B,D]
        h = self.act(self.v_proj(v) + self.c_proj(ctx).unsqueeze(1))            # [B,N,hidden]
        gate = 1.0 + self.alpha * torch.tanh(self.out(h).squeeze(-1))          # [B,N]
        return v * gate.unsqueeze(-1), gate

class XAADVR(nn.Module):
    def __init__(self, use_mr=False, use_sr=False, text_dim=512, img_dim=768, embed_dim=512, num_heads=8,
                 dropout=0.1, dvr_hidden=DVR_HIDDEN, dvr_alpha=DVR_ALPHA):
        super().__init__()
        self.use_mr, self.use_sr = bool(use_mr), bool(use_sr)
        # --- XAA cơ bản: cùng thứ tự khởi tạo với XAAFusionModule ---
        self.ref_proj = nn.Linear(img_dim, embed_dim)
        self.ref_ln   = nn.LayerNorm(embed_dim)
        self.attn = nn.MultiheadAttention(embed_dim=embed_dim, num_heads=num_heads, batch_first=True, dropout=dropout)
        self.out_ln = nn.LayerNorm(embed_dim)
        self.logit_scale = nn.Parameter(torch.tensor(math.log(1 / 0.07)))
        # --- DVR: tạo sau, rồi trả RNG về trạng thái ngay sau khi tạo XAA ---
        _rng = torch.get_rng_state()
        self.mr = ModifierGuidedRefinement(embed_dim, dvr_hidden, dvr_alpha) if self.use_mr else None
        self.sr = SelfGuidedRefinement(embed_dim, dvr_hidden, dvr_alpha) if self.use_sr else None
        torch.set_rng_state(_rng)

    def forward(self, text_hidden, text_mask, reference_hidden, return_gates=False):
        ref = self.ref_ln(self.ref_proj(reference_hidden))              # Image Projection  [B,50,512]
        gates = {}
        if self.mr is not None:
            ref, gates["mr"] = self.mr(ref, text_hidden, text_mask)    # MR
        if self.sr is not None:
            ref, gates["sr"] = self.sr(ref)                            # SR (sau MR)
        att, _ = self.attn(query=text_hidden, key=ref, value=ref)       # XAA không đổi
        x = self.out_ln(text_hidden + att)
        m = text_mask.unsqueeze(-1)
        pooled = (x * m).sum(1) / m.sum(1).clamp(min=1.0)
        out = F.normalize(pooled, dim=-1)
        return (out, gates) if return_gates else out

def query_adapter(model,reference_hidden,text_hidden,text_mask):
    with torch.inference_mode():return model(text_hidden,text_mask,reference_hidden)
```

```python
# Chỉ định nghĩa helper trong cell này; tải dữ liệu chỉ diễn ra khi chạy cell Colab bên dưới.
import contextlib
import stat
import subprocess
import tempfile
import zipfile
from pathlib import PurePosixPath

DATA_ROOT = CFG['dataset_root']
KAGGLE_DATASET = "binhhuunguyen/fashioniq"
KAGGLE_CREDENTIAL_DIR = Path("/root/.kaggle")
EXPECTED_DATASET_PATHS = {
    'captions_dir': f'{DATA_ROOT}/fashion-iq-master/fashion-iq-master/captions',
    'image_splits_dir': f'{DATA_ROOT}/fashion-iq-master/fashion-iq-master/image_splits',
    'images_dir': f'{DATA_ROOT}/resized_images/resized_images',
}
DATASET_PATHS = None
DATASET_READY = False
KAGGLE_AUTH_READY = False
_KAGGLE_USE_UPLOADED_JSON = False

class DatasetPreparationError(RuntimeError):
    pass

class DatasetAssetsMissing(DatasetPreparationError):
    pass


def _dataset_inside(path, root):
    path, root = Path(path).resolve(), Path(root).resolve()
    return path == root or root in path.parents


def _dataset_directories(root):
    """Một lượt duyệt, không đi theo symlink ra ngoài DATA_ROOT."""
    root = Path(root).resolve()
    if not root.is_dir():
        return []
    directories = []
    for directory, subdirs, filenames in os.walk(root, followlinks=False):
        subdirs[:] = sorted(d for d in subdirs
                            if not d.startswith('._') and d not in ('__MACOSX', '.kaggle_downloads')
                            and not (Path(directory)/d).is_symlink())
        if not _dataset_inside(directory, root):
            continue
        names = {name for name in filenames if not name.startswith('._')
                 and not (Path(directory)/name).is_symlink()}
        directories.append((Path(directory), names))
    return directories


def _unique_dataset_directory(candidates, label):
    candidates = sorted({str(p.resolve()) for p in candidates})
    if len(candidates) > 1:
        raise DatasetPreparationError(
            f'Nhiều thư mục phù hợp cho {label} trong DATA_ROOT; không tự chọn:\n'
            + '\n'.join(candidates))
    if not candidates:
        raise DatasetAssetsMissing(f'Chưa tìm thấy {label} đầy đủ trong DATA_ROOT.')
    return Path(candidates[0])


def _official_val_inventory(captions, splits):
    counts, anchors = {}, set()
    for category in CATEGORIES:
        cap_path = Path(captions)/f'cap.{category}.val.json'
        split_path = Path(splits)/f'split.{category}.val.json'
        if not cap_path.is_file() or not split_path.is_file():
            raise DatasetAssetsMissing(f'Thiếu official-val JSON cho category {category}.')
        try:
            triplets = json.loads(cap_path.read_text(encoding='utf-8'))
            gallery = json.loads(split_path.read_text(encoding='utf-8'))
        except (ValueError, OSError, UnicodeError):
            raise DatasetPreparationError(f'Không parse được JSON official-val của {category}.') from None
        if not isinstance(triplets, list) or not triplets:
            raise DatasetPreparationError(f'Annotation {category} phải là danh sách triplet không rỗng.')
        if not isinstance(gallery, list) or not gallery or not all(isinstance(x, str) and x for x in gallery):
            raise DatasetPreparationError(f'Image split {category} không đúng schema danh sách ID.')
        for row, triplet in enumerate(triplets):
            if not isinstance(triplet, dict):
                raise DatasetPreparationError(f'Annotation {category}, row {row}: triplet không phải object.')
            for role in ('candidate', 'target'):
                image_id = triplet.get(role)
                if (not isinstance(image_id, str) or not image_id or image_id in ('.', '..')
                        or '/' in image_id or '\\' in image_id):
                    raise DatasetPreparationError(f'Annotation {category}, row {row}: ID {role} không hợp lệ.')
        # Chỉ dùng các ID này để định vị thư mục/kiểm tra sơ bộ, không suy coverage toàn bộ.
        sample_rows = sorted({0, len(triplets)//2, len(triplets)-1})
        for row in sample_rows:
            anchors.update(triplets[row][role] for role in ('candidate', 'target'))
        counts[category] = {'query': len(triplets), 'gallery': len(gallery)}
    return counts, sorted(anchors)


def discover_dataset_paths(data_root):
    root = Path(data_root).resolve()
    directories = _dataset_directories(root)
    cap_names = {f'cap.{category}.val.json' for category in CATEGORIES}
    split_names = {f'split.{category}.val.json' for category in CATEGORIES}
    # Bao gồm cả đường dẫn encoder đã ghi nhận; không ưu tiên tùy tiện nếu có bản khác cũng phù hợp.
    caps = [directory for directory, names in directories if cap_names <= names]
    splits = [directory for directory, names in directories if split_names <= names]
    captions = _unique_dataset_directory(caps, 'captions_dir')
    image_splits = _unique_dataset_directory(splits, 'image_splits_dir')
    _, anchors = _official_val_inventory(captions, image_splits)
    jpg_names = {image_id + '.jpg' for image_id in anchors}
    candidates = [directory for directory, names in directories if jpg_names <= names]
    images = _unique_dataset_directory(candidates, 'images_dir (JPG reference/target từ annotation)')
    paths = {'captions_dir': str(captions), 'image_splits_dir': str(image_splits), 'images_dir': str(images)}
    if not all(_dataset_inside(path, root) for path in paths.values()):
        raise DatasetPreparationError('Không cho phép dùng thư mục nằm ngoài DATA_ROOT.')
    return paths


def dataset_preflight(paths, data_root, show=True):
    root = Path(data_root).resolve()
    for key in ('captions_dir', 'image_splits_dir', 'images_dir'):
        path = Path(paths[key])
        if not path.is_dir() or not _dataset_inside(path, root):
            raise DatasetAssetsMissing(f'{key} chưa tồn tại hoặc nằm ngoài DATA_ROOT.')
    counts, anchors = _official_val_inventory(paths['captions_dir'], paths['image_splits_dir'])
    images = Path(paths['images_dir'])
    # Phải dùng tên ID.jpg như image_extension hiện tại; không đổi schema/cách dựng đường dẫn benchmark.
    jpg_count = sum(p.is_file() and not p.is_symlink() and not p.name.startswith('._')
                    and p.suffix == '.jpg' for p in images.iterdir())
    if not jpg_count:
        raise DatasetAssetsMissing('Thư mục ảnh không có JPG hợp lệ về tên; bỏ qua file ._*.')
    for image_id in anchors:
        path = images/(image_id + '.jpg')
        if not path.is_file() or path.is_symlink():
            raise DatasetAssetsMissing(f'Thiếu ảnh định vị {image_id}.jpg.')
        try:
            with Image.open(path) as image:
                if image.format != 'JPEG':
                    raise ValueError('not a JPEG')
                image.load()
        except Exception:
            raise DatasetPreparationError(f'Ảnh kiểm tra sơ bộ không giải mã được: {path}') from None
    if show:
        print('Đã tìm thấy ba thư mục và parse đủ sáu file JSON official-val:')
        for category, count in counts.items():
            print(f"  {category}: {count['query']} query | {count['gallery']} gallery ID")
        for key, value in paths.items():
            print(f'  {key}: {value}')
        print(f'{jpg_count} file .jpg (không tính ._); đã giải mã {len(anchors)} ảnh định vị.')
        print('Đây là kiểm tra sơ bộ, KHÔNG xác nhận toàn bộ ảnh/coverage. Audit đầy đủ ở mục 5 vẫn bắt buộc trước metric.')
    return {'counts': counts, 'jpg_file_count': jpg_count, 'decoded_locator_images': len(anchors),
            'full_dataset_audited': False}


def ensure_kaggle_cli():
    executable=shutil.which('kaggle')
    if executable is None:
        raise EnvironmentPreflightBlocked('Kaggle CLI chưa có trong PATH. Chạy Bootstrap trước; không cài package sau khi đã import benchmark.')
    return executable


def _kaggle_command(arguments, timeout=120):
    executable = ensure_kaggle_cli()
    try:
        with contextlib.ExitStack() as stack:
            env = None  # Thử nguyên trạng credentials/env hiện có trước.
            if _KAGGLE_USE_UPLOADED_JSON:
                # Tránh token/config cũ lấn JSON mới upload; chỉ dùng một bản sao runtime tạm.
                isolated = stack.enter_context(tempfile.TemporaryDirectory(prefix='cli-', dir=KAGGLE_CREDENTIAL_DIR))
                isolated_json = Path(isolated)/'kaggle.json'
                shutil.copyfile(KAGGLE_CREDENTIAL_DIR/'kaggle.json', isolated_json)
                isolated_json.chmod(0o600)
                env = os.environ.copy()
                for key in ('KAGGLE_API_TOKEN', 'KAGGLE_USERNAME', 'KAGGLE_KEY'):
                    env.pop(key, None)
                env['KAGGLE_CONFIG_DIR'] = isolated
            completed = subprocess.run([executable, *arguments], capture_output=True, text=True,
                                       env=env, timeout=timeout, stdin=subprocess.DEVNULL)
    except subprocess.TimeoutExpired:
        raise DatasetPreparationError('Lệnh Kaggle hết thời gian; dừng trước benchmark, kiểm tra mạng rồi chạy lại.') from None
    except OSError:
        raise DatasetPreparationError('Không chạy được Kaggle CLI trong runtime hoặc chưa có JSON xác thực đã upload.') from None
    # Không in stdout/stderr của CLI: có thể chứa thông tin xác thực hoặc URL ký sẵn.
    if completed.returncode == 0:
        return 'ok'
    message = (completed.stdout or '') + '\n' + (completed.stderr or '')
    lower = message.lower()
    if any(marker in lower for marker in ('401', '403', 'unauthorized', 'forbidden', 'credential',
                                         'authenticate', 'authentication', 'kaggle.json', 'kaggle_api_token')):
        return 'auth_required'
    raise DatasetPreparationError(
        f'Kaggle CLI thất bại (exit {completed.returncode}); kiểm tra mạng/quyền truy cập dataset. '
        'Không xuất log thô để tránh lộ thông tin xác thực.') from None


def kaggle_access_status():
    return _kaggle_command(['datasets', 'files', '-d', KAGGLE_DATASET, '--page-size', '1'])


def upload_kaggle_json_if_needed():
    # Credentials chỉ nằm trong filesystem tạm của runtime, không đưa vào CFG/Drive/notebook output.
    from google.colab import files
    credential_dir = KAGGLE_CREDENTIAL_DIR
    credential_dir.mkdir(parents=True, exist_ok=True, mode=0o700)
    destination = credential_dir/'kaggle.json'
    previous_cwd = Path.cwd()
    uploaded = None
    try:
        with tempfile.TemporaryDirectory(prefix='upload-', dir=credential_dir) as private_upload:
            os.chdir(private_upload)  # files.upload() không ghi bản gốc vào Drive/current working directory.
            print('Chọn một JSON Legacy API Credentials có username và key (tên file có thể bất kỳ).')
            uploaded = files.upload()
            if not isinstance(uploaded, dict) or len(uploaded) != 1:
                raise DatasetPreparationError('Cần upload đúng một file JSON xác thực Kaggle.')
            actual_name = next(iter(uploaded))  # Dùng đúng tên files.upload() trả về, không giả định kaggle.json.
            try:
                payload = json.loads(uploaded[actual_name].decode('utf-8'))
            except (ValueError, UnicodeError, AttributeError):
                raise DatasetPreparationError('File upload không phải JSON UTF-8 hợp lệ.') from None
            if not isinstance(payload, dict) or not all(
                    isinstance(payload.get(key), str) and payload[key].strip() for key in ('username', 'key')):
                raise DatasetPreparationError('JSON cần có hai trường chuỗi không rỗng: username và key.')
            # Chỉ ghi các trường cần thiết; không lưu tên file upload hoặc credential vào artifact benchmark.
            secure_json = {'username': payload['username'].strip(), 'key': payload['key'].strip()}
            fd, temp_name = tempfile.mkstemp(prefix='.credentials-', dir=credential_dir)
            try:
                os.fchmod(fd, 0o600)
                with os.fdopen(fd, 'w', encoding='utf-8') as stream:
                    json.dump(secure_json, stream)
                os.replace(temp_name, destination)
                destination.chmod(0o600)
            finally:
                if Path(temp_name).exists():
                    Path(temp_name).unlink()
    finally:
        os.chdir(previous_cwd)
        if isinstance(uploaded, dict):
            uploaded.clear()
    print('Đã lưu JSON xác thực trong runtime với quyền 600; không in nội dung.')


def _extract_dataset_archive(archive, data_root):
    """Giải nén dưới DATA_ROOT; tiếp tục bản dở bằng cách bỏ qua file có bytes giống hệt."""
    root = Path(data_root).resolve()
    try:
        with zipfile.ZipFile(archive) as bundle:
            for member in bundle.infolist():
                relative = PurePosixPath(member.filename)
                if (relative.is_absolute() or '..' in relative.parts or '\\' in member.filename
                        or stat.S_ISLNK(member.external_attr >> 16)):
                    raise DatasetPreparationError('Archive có đường dẫn hoặc symlink không an toàn; dừng giải nén.')
                if any(part.startswith('._') or part == '__MACOSX' for part in relative.parts):
                    continue
                target = root.joinpath(*relative.parts)
                if not _dataset_inside(target, root):
                    raise DatasetPreparationError('Archive trỏ ra ngoài DATA_ROOT.')
                if member.is_dir():
                    target.mkdir(parents=True, exist_ok=True)
                    continue
                target.parent.mkdir(parents=True, exist_ok=True)
                if target.exists():
                    if not target.is_file() or target.is_symlink():
                        raise DatasetPreparationError(f'Xung đột dữ liệu có sẵn: {target}')
                    with bundle.open(member) as source:
                        expected_hash = hashlib.sha256()
                        for block in iter(lambda: source.read(8 * 1024 * 1024), b''):
                            expected_hash.update(block)
                    if file_hash(target) != expected_hash.hexdigest():
                        raise DatasetPreparationError(f'File có sẵn khác archive; không ghi đè: {target}')
                    continue
                # File .partial không được nhận là JPG/JSON đầy đủ sau khi runtime bị ngắt.
                temp = target.with_name(target.name + '.extracting')
                with bundle.open(member) as source, temp.open('wb') as output:
                    shutil.copyfileobj(source, output)
                os.replace(temp, target)
    except DatasetPreparationError:
        raise
    except Exception:
        raise DatasetPreparationError('Giải nén FashionIQ thất bại hoặc archive hỏng; không tiếp tục benchmark.') from None


def download_and_extract_dataset(data_root):
    global _KAGGLE_USE_UPLOADED_JSON
    root = Path(data_root)
    root.mkdir(parents=True, exist_ok=True)
    downloads = root/'.kaggle_downloads'
    downloads.mkdir(exist_ok=True)
    # Thư mục mới cho mỗi lần tải; không dùng ZIP dở từ lần thất bại trước.
    with tempfile.TemporaryDirectory(prefix='fashioniq-', dir=downloads) as work:
        print('Tải binhhuunguyen/fashioniq; thời gian phụ thuộc mạng và dung lượng dataset...')
        outcome = _kaggle_command(['datasets', 'download', '-d', KAGGLE_DATASET, '-p', work, '--quiet'], timeout=3600)
        if outcome == 'auth_required' and not _KAGGLE_USE_UPLOADED_JSON:
            print('Kaggle yêu cầu xác thực khi tải (listing có thể công khai); cần JSON xác thực.')
            upload_kaggle_json_if_needed()
            _KAGGLE_USE_UPLOADED_JSON = True
            outcome = _kaggle_command(['datasets', 'download', '-d', KAGGLE_DATASET, '-p', work, '--quiet'], timeout=3600)
        if outcome != 'ok':
            raise DatasetPreparationError('Kaggle từ chối tải sau xác thực; kiểm tra quyền truy cập/điều khoản dataset. Không tiếp tục benchmark.')
        archives = sorted(Path(work).glob('*.zip'))
        if len(archives) != 1 or not zipfile.is_zipfile(archives[0]):
            raise DatasetPreparationError('Tải chưa tạo đúng một ZIP hợp lệ; không giải nén/không chạy benchmark.')
        _extract_dataset_archive(archives[0], root)
```

```python
class ShardStore:
    """NPY shards + commit checksum; recovery only from a verified pending transaction. Never adopt unknown/or corrupt data."""
    def __init__(self,root,key,kind,ids,tail,dtype='float32',shard_size=None):
        self.key=key;self.kind=kind;self.ids=list(ids);self.tail=tuple(tail);self.dtype=np.dtype(dtype)
        require(self.ids and len(self.ids)==len(set(self.ids)),'Feature IDs rỗng/trùng: '+kind)
        self.id2idx={image_id:i for i,image_id in enumerate(self.ids)};self.step=shard_size or CFG['feature_shard_size']
        self.root=Path(root)/'features'/key/kind;self.root.mkdir(parents=True,exist_ok=True)
        self.local=Path(CFG['local_cache'])/key/kind;self.local.mkdir(parents=True,exist_ok=True)
        write_json(self.root/'index.json',{'fingerprint':key,'kind':kind,'ids':self.ids,'shape_tail':list(tail),'dtype':str(self.dtype),'shard_size':self.step},immutable=True)
    def read(self,start,compute=None):
        stop=min(start+self.step,len(self.ids));name=f'{start:07d}_{stop:07d}';source=self.root/(name+'.npy');commit=self.root/(name+'.json');pending=self.root/(name+'.pending.json')
        expected={'fingerprint':self.key,'kind':self.kind,'ids_sha256':fingerprint(self.ids[start:stop]),'shape':[stop-start,*self.tail],'dtype':str(self.dtype)}
        try:
            if not commit.exists():
                if not source.exists():
                    require(compute is not None,'Shard thiếu và không có hàm encode: '+str(source))
                    values=np.asarray(compute(self.ids[start:stop]));array_quality(values,self.tail,self.dtype,unit=self.kind=='target')
                    require(list(values.shape)==expected['shape'],'Encoder trả sai số sample; không reshape để che sai lệch.')
                    with tempfile.NamedTemporaryFile(dir=self.local,suffix='.npy',delete=False) as f:
                        np.save(f,values,allow_pickle=False);temporary=Path(f.name)
                    meta={**expected,'sha256':file_hash(temporary)}
                    write_json(pending,meta,immutable=True)
                    stage=source.with_name(source.name+'.'+uuid.uuid4().hex+'.partial');shutil.copyfile(temporary,stage)
                    require(not source.exists(),'Source shard xuất hiện đồng thời; dừng bảo toàn file.')
                    os.replace(stage,source);temporary.unlink()
                require(pending.is_file(),'Orphan NPY không có transaction provenance; không tự adopt/ghi đè: '+str(source))
                meta=read_json(pending)
                require(all(meta.get(k)==v for k,v in expected.items()),'Pending transaction fingerprint/ID/schema mismatch.')
                require(file_hash(source)==meta['sha256'],'Pending shard checksum mismatch; không tự ghi đè.')
                check=np.load(source,mmap_mode='r',allow_pickle=False);array_quality(check,self.tail,self.dtype,unit=self.kind=='target')
                write_json(commit,meta,immutable=True)
            meta=read_json(commit)
            require(all(meta.get(k)==v for k,v in expected.items()),'Committed shard fingerprint/ID/schema mismatch.')
            require(source.is_file() and file_hash(source)==meta['sha256'],'Shard thiếu/hỏng checksum: '+str(source))
            local=self.local/source.name
            if not local.exists() or file_hash(local)!=meta['sha256']:
                stage=local.with_name(local.name+'.tmp');shutil.copyfile(source,stage);os.replace(stage,local)
            values=np.load(local,mmap_mode='r',allow_pickle=False)
            require(list(values.shape)==expected['shape'] and values.dtype==self.dtype,'NPY shape/dtype khác commit.')
            return values
        except Exception as exc:
            write_json(self.root/(name+'.invalid.json'),{'reuse_allowed':False,'reason':f'{type(exc).__name__}: {exc}','fingerprint':self.key})
            raise CacheIntegrityError('Cache không hợp lệ; giữ nguyên dữ liệu để kiểm tra: '+str(source)) from exc
    def ensure(self,compute):
        stats=[]
        for start in tqdm(range(0,len(self.ids),self.step),desc='Encode/audit '+self.kind):stats.append(array_quality(self.read(start,compute),self.tail,self.dtype,unit=self.kind=='target'))
        report={'kind':self.kind,'fingerprint':self.key,'count':len(self.ids),'id_order_sha256':fingerprint(self.ids),'all_shards_checksum_and_values_checked':True,'shards':stats}
        write_json(self.root/'audit.json',report,immutable=True);return report
    def take(self,ids):
        indices=[self.id2idx[i] for i in ids];shards={}
        for index in indices:
            start=index//self.step*self.step
            if start not in shards:shards[start]=self.read(start)
        return np.stack([np.asarray(shards[index//self.step*self.step][index%self.step]) for index in indices])

def extraction_contract(encoder,tokenizer,processor):
    return {'implementation':PREPARATION_IMPLEMENTATION,'encoder_id':ENCODER_ID_LOCK,'revision':ENCODER_REVISION_LOCK,'encoder_weight_sha256':tensor_state_hash(encoder),
      'processor':processor.to_dict(),'tokenizer':{'class':type(tokenizer).__name__,'vocab_sha256':fingerprint(tokenizer.get_vocab()),'special_tokens':tokenizer.special_tokens_map,'max_length':77},
      'caption_policy':CAPTION_POLICY,'truncation_policy':CFG['truncation_policy'],'dtype':'float32','attention_backend':'eager','autocast':False,'tf32':False,
      'libraries':{n:importlib.metadata.version(n) for n in ('torch','transformers','tokenizers','huggingface-hub','numpy','pandas','Pillow')},
      'device':str(DEVICE),'device_name':torch.cuda.get_device_name(DEVICE) if DEVICE.type=='cuda' else platform.machine(),'cuda':torch.version.cuda,
      'batch_size':CFG['encode_batch_size'],'feature_shard_size':CFG['feature_shard_size'],
      'reference':'vision_model.last_hidden_state INCLUDING CLS; no projection or L2',
      'text':'text_model.last_hidden_state, [77,512], no projection/normalization + binary attention_mask',
      'target':'visual_projection(vision_model.pooler_output) then L2, [512]',
      'repeat_acceptance':{'atol':CFG['repeat_atol'],'rtol':CFG['repeat_rtol'],'predeclared':True,'note':'Repeat same input and execution contract; not a historical feature acceptance threshold.'}}

def extract_image_features(ids,paths,encoder,processor,role,hashes=None):
    output=[]
    for start in range(0,len(ids),CFG['encode_batch_size']):
        batch=ids[start:start+CFG['encode_batch_size']];images=[]
        for image_id in batch:
            path=Path(paths['images_dir'])/(image_id+'.jpg')
            if hashes is not None:require(file_hash(path)==hashes[image_id],'Ảnh đổi sau audit: '+image_id)
            with Image.open(path) as im:images.append(im.convert('RGB'))
        pixels=processor(images=images,return_tensors='pt')['pixel_values'].float().to(DEVICE)
        with torch.inference_mode():
            vision=encoder.vision_model(pixel_values=pixels)
            value=vision.last_hidden_state if role=='reference' else F.normalize(encoder.visual_projection(vision.pooler_output),dim=-1)
        values=value.float().cpu().numpy();array_quality(values,(50,768) if role=='reference' else (512,),'float32',unit=role=='target');output.append(values)
    return np.concatenate(output)

def extract_text_features(ids,texts,encoder,tokenizer,mask_only=False):
    output=[]
    for start in range(0,len(ids),CFG['encode_batch_size']):
        batch=ids[start:start+CFG['encode_batch_size']]
        inputs=tokenizer([texts[i] for i in batch],padding='max_length',max_length=77,truncation=False,return_tensors='pt')
        require(inputs['input_ids'].shape==(len(batch),77),'Token shape không đúng; kiểm tra tokenization audit trước encode.')
        if mask_only:value=inputs['attention_mask'].numpy().astype(np.int64)
        else:
            with torch.inference_mode():value=encoder.text_model(**{k:v.to(DEVICE) for k,v in inputs.items()}).last_hidden_state.float().cpu().numpy()
        output.append(value)
    return np.concatenate(output)

def sample_extraction_checks(q,paths,encoder,tokenizer,processor,image_rows):
    sample=pd.concat([q[q.category==c].iloc[:1] for c in CATEGORIES],ignore_index=True)
    ids=sample.reference_image_id.tolist();text=sample.set_index('query_id').modification_text.to_dict();query_ids=sample.query_id.tolist();hashes={r['image_id']:r['sha256'] for r in image_rows}
    result={'scope':'real dataset samples with real frozen encoder; NOT full feature audit or model metrics','sample_query_ids':query_ids,'sample_image_ids':ids,'comparisons':{},'passed':True}
    for kind in ('reference','target','text'):
        fn=(lambda:extract_text_features(query_ids,text,encoder,tokenizer)) if kind=='text' else (lambda kind=kind:extract_image_features(ids,paths,encoder,processor,kind,hashes))
        a,b=fn(),fn();stats=numeric_comparison(a,b);accepted=bool(np.allclose(a,b,atol=CFG['repeat_atol'],rtol=CFG['repeat_rtol']))
        result['comparisons'][kind]={'same_input_repeat':stats,'repeat_within_predeclared_tolerance':accepted}
        result['passed'] &= accepted
        if kind=='reference':
            result['comparisons']['reference_CLS']=numeric_comparison(a[:,:1],b[:,:1])
            result['comparisons']['reference_patch']=numeric_comparison(a[:,1:],b[:,1:])
            result['comparisons']['reference_per_token']=[{'token_index':i,**numeric_comparison(a[:,i],b[:,i])} for i in range(a.shape[1])]
    return result

def historical_diagnostics(q,paths,encoder,tokenizer,processor):
    report={'reuse':False,'reuse_reason':'No locked historical revision/preprocessing/caption/row provenance found in surveyed Drive assets. New heads will be initialized from scratch on one consistent new feature package.',
      'historical_encoder_revision':'unknown','historical_reproduction_claimed':False,'blocking_for_new_features':False,'comparisons':{},'mapping':{},'text_comparison':'skipped: historical text row identity not proven; never index legacy text by reset_index.',
      'limitations':['Header/sample inspection is not a full audit of legacy arrays.','Numerical drift alone neither proves a mapping bug nor establishes compatibility.','New pipeline must pass its own image, schema, repeat and shape/finite/norm audits.']}
    for role,array_name,map_name in [('reference','reference_image_hidden_states.npy','reference_id2idx.json'),('target','target_image_embeds.npy','target_id2idx.json')]:
        try:
            directory=Path(CFG['legacy_image_dir']);mp=read_json(directory/map_name)
            require(isinstance(mp,dict) and all(valid_image_id(k) and type(v) is int for k,v in mp.items()) and sorted(mp.values())==list(range(len(mp))),'Legacy image mapping không phải permutation.')
            a=np.load(directory/array_name,mmap_mode='r',allow_pickle=False)
            require(len(a)==len(mp),'Legacy mapping length mismatch.')
            ids=sorted(set(q.reference_image_id if role=='reference' else q.target_image_id)&set(mp))[:CFG['historical_sample_size']]
            report['mapping'][role]={'count':len(mp),'sha256':file_hash(directory/map_name),'array_shape':list(a.shape),'dtype':str(a.dtype),'sample_ids':ids,'full_array_audited':False}
            if ids:
                old=np.asarray(a[[mp[i] for i in ids]],dtype=np.float32);new=extract_image_features(ids,paths,encoder,processor,role)
                report['comparisons'][role]=numeric_comparison(old,new)
                if role=='reference':
                    report['comparisons']['reference_CLS']=numeric_comparison(old[:,:1],new[:,:1]);report['comparisons']['reference_patch']=numeric_comparison(old[:,1:],new[:,1:])
                    report['comparisons']['reference_per_token']=[{'token_index':i,**numeric_comparison(old[:,i],new[:,i])} for i in range(old.shape[1])]
        except Exception as exc:report['mapping'][role]={'status':'unknown_or_unusable','reason':f'{type(exc).__name__}: {exc}'}
    for name in ('fashionclip_text_hidden_states.npy','fashionclip_text_attention_masks.npy'):
        try:
            a=np.load(Path(CFG['legacy_text_dir'])/name,mmap_mode='r',allow_pickle=False)
            report.setdefault('text_headers',{})[name]={'shape':list(a.shape),'dtype':str(a.dtype),'full_array_audited':False}
        except Exception as exc:report.setdefault('text_headers',{})[name]={'status':'unavailable','reason':type(exc).__name__}
    return report

def prepare_feature_package(run,train,val,gallery,image_rows,paths,contract,encoder,tokenizer,processor):
    q=pd.concat([train,val],ignore_index=True);refs=sorted(set(q.reference_image_id));targets=sorted(set(train.target_image_id)|set(gallery.image_id))
    key=fingerprint({'contract':contract,'query_rows':q.to_dict('records'),'gallery_rows':gallery.to_dict('records'),'images':image_rows})
    pack=Path(run)/'features'/key;pack.mkdir(parents=True,exist_ok=True)
    plan={'fingerprint':key,'contract':contract,'query_order_sha256':fingerprint(q.query_id.tolist()),'reference_order_sha256':fingerprint(refs),'target_order_sha256':fingerprint(targets),
          'counts':{'text':len(q),'reference':len(refs),'target':len(targets)},'estimated_npy_payload_bytes':len(q)*(77*512*4+77*8)+len(refs)*50*768*4+len(targets)*512*4,
          'scope':'official train + official val; all required references, official train targets, full official-val galleries','legacy_arrays_mixed':False}
    write_json(pack/'feature_plan.json',plan,immutable=True)
    print('Feature package:',key[:16],'| dự kiến %.2f GiB payload NPY' % (plan['estimated_npy_payload_bytes']/1024**3))
    stores={'text':ShardStore(run,key,'text',q.query_id.tolist(),(77,512)),
            'mask':ShardStore(run,key,'mask',q.query_id.tolist(),(77,),dtype='int64'),
            'reference':ShardStore(run,key,'reference',refs,(50,768)),
            'target':ShardStore(run,key,'target',targets,(512,))}
    texts=q.set_index('query_id').modification_text.to_dict();hashes={r['image_id']:r['sha256'] for r in image_rows}
    callbacks={'text':lambda ids:extract_text_features(ids,texts,encoder,tokenizer),
               'mask':lambda ids:extract_text_features(ids,texts,encoder,tokenizer,True),
               'reference':lambda ids:extract_image_features(ids,paths,encoder,processor,'reference',hashes),
               'target':lambda ids:extract_image_features(ids,paths,encoder,processor,'target',hashes)}
    audits={kind:store.ensure(callbacks[kind]) for kind,store in stores.items()}
    for kind,store in stores.items():write_json(pack/(kind+'_id2idx.json'),store.id2idx,immutable=True)
    keys=defaultdict(list)
    for row in q.itertuples():keys[row.triplet_key_sha256].append(row.query_id)
    write_json(pack/'triplet_key_to_query_ids.json',dict(keys),immutable=True)
    return stores,key,plan,audits

def write_feature_views(run,stores,key,frames):
    output={}
    for name,frame in frames.items():
        view=frame.copy()
        if 'query_id' in view:
            view['text_feature_index']=[stores['text'].id2idx[i] for i in view.query_id]
            view['reference_feature_index']=[stores['reference'].id2idx[i] for i in view.reference_image_id]
            require(all(i in stores['target'].id2idx for i in view.target_image_id),'Target coverage thiếu trong feature view: '+name)
            view['target_feature_index']=[stores['target'].id2idx[i] for i in view.target_image_id]
        else:view['target_feature_index']=[stores['target'].id2idx[i] for i in view.image_id]
        path=Path(run)/'feature_views'/(name+'.csv');write_csv(path,view,immutable=True)
        back=read_csv(path);require(len(back)==len(frame),'Feature view readback row count mismatch.')
        column='query_id' if 'query_id' in frame else 'image_id';require(back[column].tolist()==frame[column].tolist(),'Feature view ID order mismatch.')
        output[name]={'path':str(path.relative_to(run)),'sha256':file_hash(path),'rows':len(view),'fingerprint':key}
    return output

def smoke_heads_new_features(q,stores):
    sample=pd.concat([q[q.category==c].iloc[:1] for c in CATEGORIES],ignore_index=True);ids=sample.query_id.tolist()
    rh=torch.from_numpy(stores['reference'].take(sample.reference_image_id.tolist()).copy()).to(DEVICE)
    th=torch.from_numpy(stores['text'].take(ids).copy()).to(DEVICE);mask=torch.from_numpy(stores['mask'].take(ids).copy()).to(DEVICE)
    require(not any('target' in p for p in inspect.signature(query_adapter).parameters),'Target xuất hiện trong query adapter input.')
    results={};devices=[DEVICE.index or 0] if DEVICE.type=='cuda' else []
    with torch.random.fork_rng(devices=devices):
        for name,spec in MODEL_SPECS.items():
            torch.manual_seed(42)
            model=(XAAFusionModule() if name=='xaa_noft' else XAADVR(use_mr=spec['use_mr'],use_sr=spec['use_sr'])).float().to(DEVICE).eval().requires_grad_(False)
            before=tensor_state_hash(model);batches=[]
            for batch in (1,len(ids)):
                with torch.inference_mode():native=model(th[:batch],mask[:batch],rh[:batch]);adapted=query_adapter(model,rh[:batch],th[:batch],mask[:batch])
                require(native.shape==(batch,512) and torch.isfinite(native).all(),'Head shape/finite failed: '+name)
                require(torch.allclose(native.norm(dim=-1),torch.ones(batch,device=DEVICE),atol=1e-5),'Head L2 failed: '+name)
                require(torch.allclose(native,adapted,atol=1e-6,rtol=1e-6),'Native/adapter mismatch: '+name);batches.append(batch)
            require(tensor_state_hash(model)==before,'Head weights changed in smoke test.')
            results[name]={'passed':True,'initialization':'fresh test-only random initialization, seed 42; NO training/old checkpoint loading','parameter_count':sum(p.numel() for p in model.parameters()),'batches':batches,'weights_unchanged':True,'target_input':False,'metric_computed':False}
            del model
    return results
```
